```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/House_Rent_Dataset.csv")
```

```python
# Extract the current floor number
df["Floor_Number"] = (
    df["Floor"]
    .str.extract(r"^(\d+)")
    .fillna(0)
    .astype(int)
)

# Extract the total number of floors in the building
df["Total_Floors"] = (
    df["Floor"]
    .str.extract(r"out of (\d+)")
    .fillna(0)
    .astype(int)
)
```

```python
df["Posted On"].unique()
```

```python
df["Posted On"] = pd.to_datetime(
    df["Posted On"],
    format="%d-%m-%Y"
)
df[["Posted On"]].head()
```

```python
df["Posted_Month"] = df["Posted On"].dt.month
df["Posted_Day"] = df["Posted On"].dt.day
df[["Posted On", "Posted_Month", "Posted_Day"]].head(10)
```

```python
locality_counts = df["Area Locality"].value_counts()

print("Localities appearing only once:", (locality_counts == 1).sum())
print("Localities appearing 2-5 times:", locality_counts.between(2, 5).sum())
```

## Feature Engineering Summary

- Extracted `Floor_Number` and `Total_Floors` from the `Floor` column.
- Converted `Posted On` into `Posted_Month` and `Posted_Day`.
- Removed `Area Locality` due to its high cardinality (2,235 unique localities).
- Removed the original `Floor` and `Posted On` columns after extracting useful information.
- Applied one-hot encoding to categorical features such as City, Area Type, Furnishing Status, Tenant Preferred, and Point of Contact.
- Separated the target variable `Rent` from the input features.

The final dataset contains **4,746 records and 20 input features** for rent prediction.
